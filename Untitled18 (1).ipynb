```python


# Import libraries
import pandas as pd
import numpy as np

#  Load dataset
df = pd.read_csv("/content/students.csv")

#  Display complete raw data
print("STUDENT PERFORMANCE ANALYTICS SYSTEM")
print("=" * 50)

print("\n1. COMPLETE RAW DATA")
print(df.to_string(index=False))

print("\n2. DATASET INFORMATION")

print("Total Students:", len(df))
print("Total Columns:", len(df.columns))

print("\nColumn Names:")
print(df.columns.tolist())


# Complete Student Results

subjects = ["Maths", "Python", "AI"]

# Calculate total and average
df["Total"] = df[subjects].sum(axis=1)
df["Average"] = df[subjects].mean(axis=1).round(2)

# Assign grades using a function
def calculate_grade(avg):
    if avg >= 90:
        return "A"
    elif avg >= 80:
        return "B"
    elif avg >= 70:
        return "C"
    elif avg >= 60:
        return "D"
    else:
        return "F"

df["Grade"] = df["Average"].apply(calculate_grade)

#  Pass/Fail
df["Result"] = np.where(
    (df[subjects] >= 40).all(axis=1),
    "Pass",
    "Fail"
)

# Display everything in one table
print("STUDENT PERFORMANCE REPORT")
print("=" * 75)

display(df[[
    "Student_ID", "Name", "Maths", "Python", "AI",
    "Total", "Average", "Grade", "Result"
]])


# STEP 5: Overall Class Statistics

print("CLASS PERFORMANCE STATISTICS")
print("=" * 40)

print("Total Students:", len(df))
print("Class Average:", round(df["Average"].mean(), 2))
print("Highest Average:", df["Average"].max())
print("Lowest Average:", df["Average"].min())

# Student with highest average
top_student = df.loc[df["Average"].idxmax()]

print("Best Performer:", top_student["Name"])
print("Best Performer's Average:", top_student["Average"])


# STEP 6: Subject-wise Performance Analysis

print("\nSUBJECT-WISE PERFORMANCE")
print("=" * 40)

for subject in subjects:
    print(subject, "Average:", round(df[subject].mean(), 2))

best_subject = df[subjects].mean().idxmax()
print("\nBest Performing Subject:", best_subject)


# Pass/Fail Statistics

print("\nPASS/FAIL STATISTICS")
print("=" * 40)

pass_count = (df["Result"] == "Pass").sum()
fail_count = (df["Result"] == "Fail").sum()

print("Students Passed:", pass_count)
print("Students Failed:", fail_count)


#  Top 3 Performing Students

print("\nTOP 3 PERFORMING STUDENTS")
print("=" * 40)

top_three = df.nlargest(3, "Average")

display(top_three[[
    "Student_ID", "Name", "Average", "Grade", "Result"
]])


# Final Complete Report

print("\nFINAL STUDENT PERFORMANCE REPORT")
print("=" * 70)

display(df[[
    "Student_ID", "Name", "Department",
    "Maths", "Python", "AI", "Attendance",
    "Total", "Average", "Grade", "Result"
]])

```
